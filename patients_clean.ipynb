```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('patients_dirty.csv')
```

```python
df_patients = df.copy()
```

```python
df_patients = df_patients.drop_duplicates()
df_patients = df_patients.reset_index(drop=True)
```

```python
df_patients['first_name'] = df_patients['first_name'].str.strip()
df_patients['first_name'] = df_patients['first_name'].str.upper()

df_patients['first_name'] = np.where(df_patients['first_name'] == '-', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == '?', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == 'N/A', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == 'NONE', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == 'NULL', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == 'UNKNOWN', np.nan, df_patients['first_name'])
df_patients['first_name'] = np.where(df_patients['first_name'] == '', np.nan, df_patients['first_name'])

df_patients['first_name'] = df_patients['first_name'].str.title()
```

```python
df_patients['last_name'] = df_patients['last_name'].str.strip()
df_patients['last_name'] = df_patients['last_name'].str.upper()

df_patients['last_name'] = np.where(df_patients['last_name'] == '-', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == '?', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == 'N/A', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == 'NONE', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == 'NULL', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == 'UNKNOWN', np.nan, df_patients['last_name'])
df_patients['last_name'] = np.where(df_patients['last_name'] == '', np.nan, df_patients['last_name'])

df_patients['last_name'] = df_patients['last_name'].str.title()
```

```python
df_patients['birth_date'] = df_patients['birth_date'].str.strip()
df_patients['birth_date'] = pd.to_datetime(df_patients['birth_date'], format='mixed', errors='coerce')
```
